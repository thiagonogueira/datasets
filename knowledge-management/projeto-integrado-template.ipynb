### 2.1. Recuperação

```python
def medir_recuperacao(trechos_por_chamado, rotulo):
    """Fração dos chamados anotados em que todos os trechos de referência aparecem nos trechos recuperados.
    trechos_por_chamado: dicionário id_reclamacao -> lista de trechos recuperados.
    Devolve acertos, total e a lista dos chamados errados."""
    anotados = [ch for ch in chamados if ch["trecho_referencia"]]
    acertos, chamados_errados = 0, []
    for ch in anotados:
        texto = "\n".join(trechos_por_chamado[ch["id_reclamacao"]]).lower()
        partes = ch["trecho_referencia"].split(" [...] ")
        if all(p.lower() in texto for p in partes):
            acertos += 1
        else:
            chamados_errados.append(ch["id_reclamacao"])
    print(f"{rotulo}. recuperação {acertos} de {len(anotados)}")
    return acertos, len(anotados), chamados_errados
```
